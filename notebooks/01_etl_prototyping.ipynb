# ETL Prototyping: Commercial Data Cleaning
**Objective:** Read legacy financial spreadsheets, clean currency/percentage strings, and standardize the schema for PostgreSQL ingestion.

```python
import pandas as pd
```

## 1. Data Ingestion
Loading the raw mock data generated in the previous step.

```python
# Load raw legacy data
df_annual = pd.read_csv('../data/metas_anuais.csv')
df_monthly = pd.read_csv('../data/metas_mensais.csv')
df_sellers = pd.read_csv('../data/metas_vendedores.csv')
```

## 2. Cleaning Functions
Creating standard functions to strip Brazilian currency symbols ('R$'), percentage signs ('%'), and fix decimal separators for float conversion.

```python
def clean_currency(value_str):
    if pd.isna(value_str):
        return 0.0
    # Remove 'R$', spaces, and thousand separators (dots), then replace decimal comma with dot
    cleaned = str(value_str).replace('R$','').replace(' ','').replace('.','').replace(',','.')
    return float(cleaned)

def clean_percentage(value_str):
    if pd.isna(value_str):
        return 0.0
    cleaned = str(value_str).replace('%','').replace(',','.')
    return float(cleaned)
```

## 3. Transformation & Schema Standardization (Annual Targets)
Applying the cleaning functions and translating legacy Portuguese columns to English snake_case.

```python
df_annual["totalPrevistoAno"] = df_annual["totalPrevistoAno"].apply(clean_currency)
df_annual["totalRealizadoAno"] = df_annual["totalRealizadoAno"].apply(clean_currency)
df_annual["metaConcluidaPorcentagem"] = df_annual["metaConcluidaPorcentagem"].apply(clean_percentage)

annual_mapping = {
    'id': 'source_id',
    'tipo': 'category',
    'totalPrevistoAno': 'predicted_total_year',
    'totalRealizadoAno': 'realized_total_year',
    'metaConcluidaPorcentagem': 'completion_percentage'
}

df_annual = df_annual.rename(columns = annual_mapping)

print("Annual Targets (Cleaned)")
display(df_annual.head())
```

## 4. Transformation & Schema Standardization (Monthly Targets & Sellers)
Applying the same robust transformations to the monthly aggregates and seller performance data.

```python
df_monthly['previstoMes'] = df_monthly['previstoMes'].apply(clean_currency)
df_monthly['realizadoMes'] = df_monthly['realizadoMes'].apply(clean_currency)
df_monthly['concluidoMesPorcentagem'] = df_monthly['concluidoMesPorcentagem'].apply(clean_percentage)

monthly_mapping = {
    'id': 'source_id',
    'tipo': 'category',
    'previstoMes': 'predicted_total_month',
    'realizadoMes': 'realized_total_month',
    'concluidoMesPorcentagem': 'completion_percentage',
    'mes': 'month_name',
    'nMes': 'month_number'
}

df_monthly = df_monthly.rename(columns = monthly_mapping)
df_monthly['month_number'] = df_monthly['month_number'].astype(int) # Ensure integer

# 4.2 Clean and map Seller Performance
df_sellers['concluidoMesPorcentagem'] = df_sellers['concluidoMesPorcentagem'].apply(clean_percentage)

seller_mapping = {
    'id': 'source_id',
    'vendedor': 'seller_name',
    'concluidoMesPorcentagem': 'completion_percentage',
    'mes': 'month_name',
    'nMes': 'month_number'
}

df_sellers = df_sellers.rename(columns = seller_mapping)
df_sellers['month_number'] = df_sellers['month_number'].astype(int)

print("Monthly Targets (Cleaned):")
display(df_monthly.head())

print("\nSeller Performance (Cleaned):")
display(df_sellers.head())
```
