# Mock Data Generation: Commercial Pipeline
**Objective:** Create synthetic, "dirty" data representing legacy commercial spreadsheets to simulate a real-world ETL pipeline extraction environment.

```python
import pandas as pd
import random
import os

os.makedirs('../data', exist_ok=True)

#Base setup for generation
months = ['Janeiro', 'Fevereiro','Março','Abril']
sellers = ['Gabriel Cruz', 'Lucas Silva', 'Mariana Costa']
```

## 1. Annual Targets Data
Simulating high-level annual financial targets with dirty strings (currency symbols and percentage signs).

```python
annual_data = {
    'id': [1,2],
    'tipo': ['Venda Direta','Serviços'],
    'totalPrevistoAno': ['R$ 1.500.000,00', 'R$ 800.000,00'],
    'totalRealizadoAno': ['R$ 1.250.000,50', 'R$ 850.000,00'],
    'metaConcluidaPorcentagem': ['83,33%', '106,25%']
}
pd.DataFrame(annual_data).to_csv('../data/metas_anuais.csv', index=False)
print("Annual targets generated.")
```

## 2. Monthly Targets Data
Generating dynamic monthly targets with randomized realization rates and Brazilian currency formatting (commas for decimals).

```python
monthly_data = []
id_counter = 1

for n_month, month in enumerate(months, start=1):
    for category in ['Venda Direta','Serviços']:
        predicted = random.uniform(50000, 150000)
        realized = predicted * random.uniform(0.7, 1.2) # Varies from 70% to 120% of target
        perc = (realized / predicted) * 100

        monthly_data.append({
            'id': id_counter,
            'tipo': category,
            'previstoMes': f'R$ {predicted:,.2f}'.replace(',','X').replace('.',',').replace('X','.'),
            'realizadoMes': f'R$ {realized:,.2f}'.replace(',','X').replace('.',',').replace('X','.'),
            'mes': month,
            'nMes': str(n_month) # String to simulate TS legacy
        })
        id_counter += 1

pd.DataFrame(monthly_data).to_csv('../data/metas_mensais.csv', index=False)

print("Monthly targets generated.")
```

## 3. Seller Performance Data
Generating performance metrics for individual sellers across the recorded months.

```python
seller_data = []
id_counter = 1

for n_month, month in enumerate(months, start=1):
    for seller in sellers:
        perc = random.uniform(60,110)
        seller_data.append({
            'id': id_counter,
            'vendedor': seller,
            'concluidoMesPorcentagem': f'{perc:.1f}%'.replace('.',','),
            'mes': month,
            'nMes': str(n_month)
        })

pd.DataFrame(seller_data).to_csv('../data/metas_vendedores.csv',index=False)

print("✅ All synthetic CSV files generated successfully in the 'data/' folder!")
```
